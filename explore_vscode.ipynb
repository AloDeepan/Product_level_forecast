```python
# Get unique order dates count for each family
cur.execute("""
SELECT
    families,
    COUNT(DISTINCT order_date) as unique_order_dates_count
FROM gold.gold_revenue_by_planning_category
GROUP BY families
ORDER BY unique_order_dates_count DESC
""")

unique_dates_by_family = cur.fetch_dataframe()
print(unique_dates_by_family)
```
